```python
import pandas as pd
import numpy as np;
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.countplot(x='Category', data=df)

plt.title('Number of Transactions in Each Category')
plt.xlabel('Category')
plt.ylabel('Number of Transactions')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(x='TransactionType', data=df)

plt.title('Number of Transactions by Transaction Type')
plt.xlabel('Transaction Type')
plt.ylabel('Number of Transactions')
plt.show()
```

```python
amount = df.groupby('TransactionType')['Amount'].sum().reset_index()

sns.barplot(x='TransactionType', y='Amount', data=amount)

plt.title('Debit vs Credit Amount Comparison')
plt.xlabel('Transaction Type')
plt.ylabel('Total Amount')
plt.show()
```

```python
spending = df.groupby('CustomerSegment')['Amount'].sum().reset_index()

sns.barplot(x='CustomerSegment', y='Amount', data=spending)

plt.title('Spending Pattern by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Amount')
plt.show()
```

```python
payment_count = df['PaymentMode'].value_counts()

plt.pie(payment_count,
        labels=payment_count.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title('Payment Mode Usage Distribution')
plt.show()
```

```python
sns.boxplot(x='Category', y='Amount', data=df)

plt.title('Transaction Amount by Category')
plt.xlabel('Category')
plt.ylabel('Transaction Amount')
plt.xticks(rotation=45)
plt.show()
```
